```python
arr3 = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def prime_return(arr):
    primes=[]
    for num in arr.flat:
        if num==2:
            continue
        is_prime=True
        for i in range (2, num):
            if num%i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
    return np.array(primes)

print(prime_return(arr3))
```

```python
def zeros_array(a,b):
    array = np.zeros((a,b),dtype=int)
    return array 
print(zeros_array(8,8))

print()

def board_array(a,b):
    array = np.zeros((a,b),dtype=int)
    array[::2, ::2]=1
    array[1::2, 1::2]=1
    return array 
print(board_array(8,8))

print()

def reverse_array(a,b):
    array = np.zeros((a,b),dtype=int)
    array[::2, 1::2]=1
    array[1::2, ::2]=1
    return array 
print(reverse_array(8,8))
```

```python
universe = np.array(['galaxy','clusters'])

def big_bang(array,spaces):
    seperator = ' ' * spaces
    return np.char.join(seperator, array)    

print(big_bang(universe,3))
```
