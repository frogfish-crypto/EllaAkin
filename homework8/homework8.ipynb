```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    y = A*np.sin(w*x)
    return y

x = np.linspace(0, 2*np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
fig, axs = plt.subplots(1, 3, figsize=(25, 5))

for A, w in zip(amplitude, frequency):
    y1 = make_sine_wave(x, A, 1)
    axs[0].plot(x, y1)
    y2 = make_sine_wave(x, 1, w)
    axs[1].plot(x, y2)
    y3 = make_sine_wave(x, A, w)
    axs[2].plot(x, y3)

axs[0].set_title("Amplitude Varies (w = 1)", fontsize=16)
axs[1].set_title("Frequency Varies (A = 1)", fontsize=16)
axs[2].set_title("Amplitude & Frequency Vary", fontsize=16)

plt.show()
```
