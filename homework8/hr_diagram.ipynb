```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("star_data.txt", skiprows=1)
print(data.shape)
print("Rows:", data.shape[0])
magnitude = data[:, 0]
temperature = data[:, 1]

print("Min magnitude:", np.min(magnitude), ",", "Star row index:", np.argmin(magnitude))
print("Max magnitude:", np.max(magnitude), ",", "Star row index:", np.argmax(magnitude))
#the brightest star is at index 1988, the dimmest star is at index 264
print("Min temperature:", np.min(temperature), "Star row index:", np.argmin(temperature))
print("Max temperature:", np.max(temperature), ",", "Star row index:", np.argmax(temperature))

log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)



brightest_idx = np.argmin(magnitude)   
dimmest_idx = np.argmax(magnitude)  
hottest_idx = np.argmax(temperature)
coldest_idx = np.argmin(temperature)

fig, axs = plt.subplots(1,1, figsize=(8,6))

plt.scatter(log_temperature, log_magnitude, label="Stars")


plt.scatter(log_temperature[brightest_idx], log_magnitude[brightest_idx],
            color="black", s=200, label="Brightest Star")
plt.scatter(log_temperature[dimmest_idx], log_magnitude[dimmest_idx],
            color="purple", s=200, label="Dimmest Star")
plt.scatter(log_temperature[hottest_idx], log_magnitude[hottest_idx],
            color="green", s=200, label="Hottest Star")
plt.scatter(log_temperature[coldest_idx], log_magnitude[coldest_idx],
            color="orange", s=200, label="Coldest Star")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()


plt.xlabel("Log Temperature")
plt.ylabel("Log Magnitude")
plt.title("Hertzsprung-Russell (HR) Diagram")
plt.legend()
plt.show()
```
