```python
import numpy as np
```

```python
my_list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(my_list)
multiply = arr * 4
print(multiply)
print("Mean:",np.mean(arr))
print("Median:",np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Index of Smallest Value:", np.argmin(arr))
print("Index of Largest Value:", np.argmax(arr))
new_arr=arr[arr>20]
print(new_arr)
```
